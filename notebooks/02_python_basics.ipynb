```python
sentence = "Community leaders support community-based adaptation planning in many communities."
print(sentence)
print(type(sentence))
```

```python
lower_sentence = sentence.lower()
words = lower_sentence.split()
print(words)
print(len(words))
```

```python
print(len(sentence))
```

```python
counts = {}
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
print(counts)
```

```python
import re

clean_words = re.findall(r"[a-z]+", sentence.lower())
print(clean_words)
```

```python
words_keep_hyphens = re.findall(r"[a-z]+(?:-[a-z]+)*", sentence.lower())
print(words_keep_hyphens)
```

```python
def tokenize(text):
    return re.findall(r"[a-z]+(?:-[a-z]+)*", text.lower())

def contains_keyword(text, keyword):
    return keyword in tokenize(text)

print(tokenize(sentence))
print(contains_keyword(sentence, "community"))
print(contains_keyword(sentence, "community-based"))
print(contains_keyword(sentence, "communities"))
```

```python
def count_words(word_list):
    counts = {}
    for word in word_list:
        if word in counts:
            counts[word] = counts[word] + 1
        else:
            counts[word] = 1
    return counts

result = count_words(tokenize (sentence))
print(result)
```

## Cleaning choices

- Text is lowercased and split into words with a regular expression.
- Hyphenated words are kept together (for example "community-based"), because
  "community-based adaptation" is a standard phrase and splitting it would hide it.
- The cleaning drops numbers, apostrophes and accented letters. This is a known
  limit. It could affect words like "women's" and non-English names.
- "community", "communities" and "community-based" are counted as separate words.
  I will decide how to group them in the keyword list.
